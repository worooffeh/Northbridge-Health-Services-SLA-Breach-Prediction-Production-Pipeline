## The challenge
A good offline metric is not a production system. The model must be verifiable, explainable, observable and replaceable when the business process changes.

## Champion lifecycle
1. Save the chosen preprocessing + model as one bundle.
2. Log metrics and metadata to MLflow.
3. Explain the strongest drivers using SHAP or interpretable coefficients.
4. Serve `/health`, `/verify` and `/predict` through FastAPI.
5. Display experiment and drift status in Dash.
6. Monitor feature PSI and later outcome metrics.
7. Retrain only after a controlled chronological challenger beats the current champion.

## Why drift is not one number
Input drift says the incoming population changed. Performance drift says the relationship between features and breach changed. Calibration drift says the probabilities no longer mean what they used to mean. These require different responses: investigate, recalibrate, retrain, or redesign features.

## Production success
Success is not “the API returns 200”. It means:
- the exact champion artifact is verified;
- online features obey the same definitions as training;
- risk scores remain calibrated and useful for intervention;
- drift alerts lead to a governed retraining decision.
